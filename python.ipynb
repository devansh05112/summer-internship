```python
a=int(input("enter first number-> "))
b=int(input("enter second number-> "))
op = input("enter operator(+,-,/,*)-> ")
if op == "+" :
  print(a+b)
elif op == "-":
  print(a-b)
elif op == "*":
  print(a*b)
elif op == "/":
  if b == 0:
    print("division error")
  else:
    print(a/b)
else:
  print("invalid input")
```
